```python
"""KNN examples using a small Vijay movie dataset.

The numeric features are an educational dataset created for this demo, not
official box-office or review statistics.
"""

from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from sklearn.metrics import accuracy_score, mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier, KNeighborsRegressor, NearestNeighbors
from sklearn.preprocessing import StandardScaler

RANDOM_STATE = 42
OUTPUT_DIR = Path('.') / "vijay_knn_outputs" # Corrected path definition
FEATURES = [
    "year",
    "runtime_min",
    "action_score",
    "family_score",
    "music_score",
    "mass_score",
]
```

```python
def make_vijay_data():
    """Return illustrative features for selected Vijay movies."""
    return pd.DataFrame([
        [2004, "Ghilli", 160, 9, 6, 8, 4, 8.1, 1],
        [2007, "Pokkiri", 165, 9, 4, 7, 4, 7.5, 1],
        [2012, "Thuppakki", 165, 9, 4, 7, 5, 8.0, 1],
        [2013, "Thalaivaa", 182, 8, 5, 6, 5, 6.2, 0],
        [2014, "Jilla", 176, 8, 7, 7, 6, 6.8, 0],
        [2014, "Kaththi", 166, 8, 6, 7, 6, 8.0, 1],
        [2016, "Theri", 157, 8, 7, 8, 7, 7.3, 1],
        [2017, "Mersal", 169, 8, 7, 8, 8, 7.6, 1],
        [2018, "Sarkar", 163, 8, 5, 6, 6, 6.8, 1],
        [2019, "Bigil", 179, 8, 8, 8, 7, 7.0, 1],
        [2021, "Master", 179, 8, 6, 6, 7, 7.3, 1],
        [2022, "Beast", 155, 9, 4, 5, 6, 5.2, 0],
        [2023, "Varisu", 169, 6, 9, 9, 7, 6.0, 1],
        [2023, "Leo", 164, 10, 4, 6, 7, 7.2, 1],
        [2024, "The GOAT", 183, 9, 5, 6, 7, 6.0, 1],
    ], columns=["year", "movie", "runtime_min", "action_score", "family_score", "music_score", "mass_score", "rating", "hit"])


def scale_data(data):
    scaler = StandardScaler()
    return scaler.fit_transform(data[FEATURES]), scaler


def show_table(title, table):
    """Display a DataFrame in Colab/Jupyter and print it in a terminal."""
    print(f"\n{title}")
    try:
        from IPython.display import display

        display(table)
    except ImportError:
        print(table.to_string(index=False))


def add_table(axis, title, table, font_size=9):
    """Add a compact pandas table to the combined results page."""
    axis.axis("off")
    axis.set_title(title, loc="left", fontsize=12, fontweight="bold", pad=8)
    chart_table = axis.table(
        cellText=table.values,
        colLabels=table.columns,
        loc="center",
        cellLoc="center",
    )
    chart_table.auto_set_font_size(False)
    chart_table.set_fontsize(font_size)
    chart_table.scale(1, 1.45)
    for cell in chart_table.get_celld().values():
        cell.set_edgecolor("#d9e2e8")
    for (row, _), cell in chart_table.get_celld().items():
        if row == 0:
            cell.set_facecolor("#264653")
            cell.get_text().set_color("white")
            cell.get_text().set_weight("bold")


def main():
    OUTPUT_DIR.mkdir(exist_ok=True)
    movies = make_vijay_data()
    X_scaled, scaler = scale_data(movies)
    movies.to_csv(OUTPUT_DIR / "vijay_movie_data.csv", index=False)

    print("=" * 78)
    print("VIJAY MOVIE KNN DEMO")
    print("=" * 78)
    print("The scores are illustrative values for learning KNN, not official statistics.")
    print("\nSTEP 1: MOVIE DATA")
    print(movies.to_string(index=False))

    print("\nSTEP 2: FEATURES USED BY KNN")
    print(f"Features: {', '.join(FEATURES)}\n")
    print("All numeric features are standardized before distance calculations.")

    # Model 1: classify whether a movie is in the illustrative hit group.
    train_idx, test_idx = train_test_split(
        movies.index, test_size=0.27, random_state=RANDOM_STATE, stratify=movies["hit"]
    )
    classifier = KNeighborsClassifier(n_neighbors=3, weights="distance")
    classifier.fit(X_scaled[train_idx], movies.loc[train_idx, "hit"])
    hit_predictions = classifier.predict(X_scaled[test_idx])
    print("\nSTEP 3: MODEL 1 - KNN CLASSIFIER (HIT / NOT HIT)")
    classifier_output = pd.DataFrame({
        "movie": movies.loc[test_idx, "movie"].values,
        "actual": movies.loc[test_idx, "hit"].map({0: "NOT HIT", 1: "HIT"}).values,
        "predicted": pd.Series(hit_predictions).map({0: "NOT HIT", 1: "HIT"}).values,
    })
    classifier_output.to_csv(OUTPUT_DIR / "model_1_classifier_output.csv", index=False)
    show_table(
        f"Accuracy: {accuracy_score(movies.loc[test_idx, 'hit'], hit_predictions):.2f}",
        classifier_output,
    )

    # Model 2: predict the illustrative rating from neighboring movies.
    regressor = KNeighborsRegressor(n_neighbors=3, weights="distance")
    regressor.fit(X_scaled[train_idx], movies.loc[train_idx, "rating"])
    rating_predictions = regressor.predict(X_scaled[test_idx])
    print("\nSTEP 4: MODEL 2 - KNN REGRESSOR (PREDICT RATING)")
    regressor_output = pd.DataFrame({
        "movie": movies.loc[test_idx, "movie"].values,
        "actual_rating": movies.loc[test_idx, "rating"].values,
        "predicted_rating": rating_predictions,
    }).round(2)
    regressor_output.to_csv(OUTPUT_DIR / "model_2_regressor_output.csv", index=False)
    show_table(
        f"Mean absolute error: {mean_absolute_error(movies.loc[test_idx, 'rating'], rating_predictions):.2f}",
        regressor_output,
    )
    rmse = mean_squared_error(movies.loc[test_idx, "rating"], rating_predictions) ** 0.5
    r2 = r2_score(movies.loc[test_idx, "rating"], rating_predictions)
    print(f"RMSE: {rmse:.2f}")
    print(f"R2 score: {r2:.2f}")

    # Fit on all available movies for a car-price-style prediction of a new movie.
    final_rating_model = KNeighborsRegressor(n_neighbors=3, weights="distance")
    final_rating_model.fit(X_scaled, movies["rating"])
    new_movie = pd.DataFrame([
        {
            "year": 2025,
            "runtime_min": 170,
            "action_score": 9,
            "family_score": 7,
            "music_score": 8,
            "mass_score": 8,
        }
    ])
    new_movie_scaled = scaler.transform(new_movie[FEATURES])
    predicted_rating = final_rating_model.predict(new_movie_scaled)[0]
    prediction_details = new_movie.copy()
    prediction_details["predicted_rating"] = round(predicted_rating, 2)
    prediction_details.to_csv(OUTPUT_DIR / "new_movie_rating_prediction.csv", index=False)
    show_table("New Vijay movie rating prediction", prediction_details)

    with (OUTPUT_DIR / "vijay_rating_model_report.txt").open("w", encoding="utf-8") as report:
        report.write("VIJAY MOVIE KNN RATING PREDICTION\n")
        report.write("=================================\n\n")
        report.write("Target: illustrative movie rating\n")
        report.write(f"Features: {', '.join(FEATURES)}\n")
        report.write(f"MAE: {mean_absolute_error(movies.loc[test_idx, 'rating'], rating_predictions):.2f}\n")
        report.write(f"RMSE: {rmse:.2f}\n")
        report.write(f"R2: {r2:.2f}\n")
        report.write(f"New movie predicted rating: {predicted_rating:.2f}\n")

    # Model 3: find movies with the closest feature profiles.
    recommender = NearestNeighbors(n_neighbors=4, metric="euclidean")
    recommender.fit(X_scaled)
    print("\nSTEP 5: MODEL 3 - KNN RECOMMENDER (SIMILAR MOVIES)")
    recommendation_rows = []
    for movie_name in ["Thuppakki", "Mersal", "Leo"]:
        movie_index = movies.index[movies["movie"] == movie_name][0]
        distances, indices = recommender.kneighbors(X_scaled[[movie_index]])
        for distance, neighbor_index in zip(distances[0][1:], indices[0][1:]):
            similarity = 1 / (1 + distance)
            recommendation_rows.append({
                "watched_movie": movie_name,
                "recommended_movie": movies.loc[neighbor_index, "movie"],
                "similarity": round(similarity, 2),
            })
    recommendation_output = pd.DataFrame(recommendation_rows)
    recommendation_output.to_csv(OUTPUT_DIR / "model_3_recommender_output.csv", index=False)
    show_table("Similar movie recommendations", recommendation_output)

    print("\nSTEP 6: ALL MODEL OUTPUT ON ONE PAGE")
    dashboard = plt.figure(figsize=(16, 14), facecolor="white")
    dashboard.suptitle(
        "Vijay Movie KNN Model Results",
        fontsize=20,
        fontweight="bold",
        color="#264653",
    )
    dashboard.text(
        0.5,
        0.955,
        "Classifier | Rating prediction | New movie prediction | Recommendations",
        ha="center",
        fontsize=11,
        color="#5c6770",
    )
    dashboard_grid = dashboard.add_gridspec(3, 2, top=0.92, hspace=0.35, wspace=0.15)
    add_table(dashboard.add_subplot(dashboard_grid[0, 0]), "Model 1: Hit classification", classifier_output)
    add_table(dashboard.add_subplot(dashboard_grid[0, 1]), "Model 2: Rating prediction", regressor_output)
    add_table(dashboard.add_subplot(dashboard_grid[1, 0]), "New movie prediction", prediction_details)
    add_table(dashboard.add_subplot(dashboard_grid[1, 1]), "Model 3: Similar movies", recommendation_output, font_size=8)
    metrics_axis = dashboard.add_subplot(dashboard_grid[2, :])
    metrics_axis.axis("off")
    metrics_axis.text(
        0.5,
        0.65,
        f"Classifier accuracy: {accuracy_score(movies.loc[test_idx, 'hit'], hit_predictions):.2f}    |    "
        f"Rating MAE: {mean_absolute_error(movies.loc[test_idx, 'rating'], rating_predictions):.2f}    |    "
        f"RMSE: {rmse:.2f}    |    R2: {r2:.2f}",
        ha="center",
        fontsize=14,
        color="#264653",
    )
    metrics_axis.text(
        0.5,
        0.35,
        "Note: scores and ratings are illustrative educational data, not official movie statistics.",
        ha="center",
        fontsize=10,
        color="#5c6770",
    )
    dashboard_path = OUTPUT_DIR / "vijay_knn_all_outputs_dashboard.png"
    dashboard.savefig(dashboard_path, dpi=150, bbox_inches="tight")
    print(f"All model output saved on one page: {dashboard_path}")

    print("\nSTEP 7: GRAPHICAL OUTPUT")
    fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
    axes[0].scatter(movies["year"], movies["rating"], c=movies["hit"], cmap="RdYlGn", s=90)
    for _, row in movies.iterrows():
        axes[0].annotate(row["movie"], (row["year"], row["rating"]), fontsize=8, xytext=(3, 3), textcoords="offset points")
    axes[0].set_title("Vijay movies: year vs rating")
    axes[0].set_xlabel("Release year")
    axes[0].set_ylabel("Illustrative rating")
    axes[0].grid(alpha=0.25)

    model_names = ["Classifier", "Regressor", "Recommender"]
    model_values = [accuracy_score(movies.loc[test_idx, "hit"], hit_predictions),
                    1 - mean_absolute_error(movies.loc[test_idx, "rating"], rating_predictions) / 3,
                    1.0]
    axes[1].bar(model_names, model_values, color=["#264653", "#2a9d8f", "#e9c46a"])
    axes[1].set_title("KNN model output overview")
    axes[1].set_ylabel("Score / normalized value")
    axes[1].set_ylim(0, 1.1)
    axes[1].tick_params(axis="x", rotation=15)
    fig.tight_layout()
    figure_path = OUTPUT_DIR / "vijay_knn_results.png"
    fig.savefig(figure_path, dpi=150)

    plt.figure(figsize=(6, 5))
    plt.scatter(movies.loc[test_idx, "rating"], rating_predictions, color="#e76f51", s=90)
    plt.plot([5, 8.5], [5, 8.5], "--", color="#264653", label="Perfect prediction")
    plt.xlabel("Actual rating")
    plt.ylabel("Predicted rating")
    plt.title("KNN rating prediction: actual vs predicted")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "vijay_rating_actual_vs_predicted.png", dpi=150)
    print(f"Graph saved to: {figure_path}")
    plt.show()

    return movies, X_scaled, scaler, train_idx, test_idx, hit_predictions, rating_predictions, rmse, r2, classifier_output, regressor_output, prediction_details, recommendation_output, final_rating_model, OUTPUT_DIR
```

```python
from sklearn.naive_bayes import BernoulliNB, MultinomialNB, GaussianNB

print("\nSTEP 8: COMPARING NAIVE BAYES CLASSIFIERS")

# Initialize and train Bernoulli Naive Bayes
bernoulli_nb = BernoulliNB()
bernoulli_nb.fit(X_scaled[train_idx], movies.loc[train_idx, 'hit'])
bernoulli_predictions = bernoulli_nb.predict(X_scaled[test_idx])
bernoulli_accuracy = accuracy_score(movies.loc[test_idx, 'hit'], bernoulli_predictions)
print(f"Bernoulli Naive Bayes Accuracy: {bernoulli_accuracy:.2f}")

# Initialize and train Gaussian Naive Bayes
gaussian_nb = GaussianNB()
gaussian_nb.fit(X_scaled[train_idx], movies.loc[train_idx, 'hit'])
gaussian_predictions = gaussian_nb.predict(X_scaled[test_idx])
gaussian_accuracy = accuracy_score(movies.loc[test_idx, 'hit'], gaussian_predictions)
print(f"Gaussian Naive Bayes Accuracy: {gaussian_accuracy:.2f}")

# Multinomial Naive Bayes usually expects non-negative integer counts.
# Since our features are scaled (can be negative), we'll try to convert them to positive values
# or use a different scaler if appropriate for MultinomialNB.
# For this demonstration, we'll shift the scaled data to be non-negative.
# In a real scenario, consider if MultinomialNB is truly appropriate for such data.
X_scaled_positive = X_scaled - X_scaled.min()

multinomial_nb = MultinomialNB()
multinomial_nb.fit(X_scaled_positive[train_idx], movies.loc[train_idx, 'hit'])
multinomial_predictions = multinomial_nb.predict(X_scaled_positive[test_idx])
multinomial_accuracy = accuracy_score(movies.loc[test_idx, 'hit'], multinomial_predictions)
print(f"Multinomial Naive Bayes Accuracy: {multinomial_accuracy:.2f}")

# Create a DataFrame for comparison
nb_comparison = pd.DataFrame({
    'Model': ['BernoulliNB', 'MultinomialNB', 'GaussianNB'],
    'Accuracy': [bernoulli_accuracy, multinomial_accuracy, gaussian_accuracy]
})
show_table("Naive Bayes Classifier Comparison", nb_comparison)
```
